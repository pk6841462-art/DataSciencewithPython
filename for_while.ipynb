```python
#print numbers from 1 to 10
n=int(input("enter limit"))
for a in range(1,n,1):
    print(a)
```

```python
#print all even numbers till given number
num=int(input("enter number:"))
for a in range(1,num+1):
    if a%2==0:
        print(a,end=" ")
   
```

```python
#sum of numbers from 1 to 50
x=int(input("enter the limit"))
sum=0
for a in range(1,x+1):
    print(a,end=" ")
    sum=sum+a
print("total:",sum)
```

```python
#print table of a number
number=int(input("enter any number to print its table:"))
for a in range(1,number+1):
    print(number,"x",a,"=",number*a)
```

```python
#count numbers divisible by 3 between 1 to 100
a=100
count=0
for i in range(1,a,1):
    if i%3==0:
        count+=1
        print(i,end=" ")
print()
print("total numbers divisible by 3:",count)
```

```python
#print number 1 to 10 using while loop
i=1
while i<=10:
    print(i,end=" ")
    i=i+1
```

```python
l=int(input("enter limit:"))
i=1
while i<=l:
    if i%3==0:
        print(i,end=" ")
    i=i+1
```

```python
#print reverse
i=int(input("enter number:"))
while i>=1:
    print(i,end=" ")
    i=i-1
```

```python
#find factorial
f=int(input("enter any number:"))
fact=1
i=1
while i<=f:
    fact=fact*i
    i=i+1
print("factorial:",fact)
```
