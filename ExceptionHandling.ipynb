```python
''' exception is an unwanted situation that occurs during runtime or at the time of execution.
steps taken to make program execution in smooth manner we use alternate statements where exception occurs.
it is called exception handling'''
```

```python
'''alternate statement'''
a=int(input())
b=int(input())
if b==0:
    b=2
ans=a/b
print(ans)
```

```python
'''in exception handling we find risky code in the program, code where exception may occurs is called risky code.
the risky code goes in the try block and exception in except block'''
# try except block

a=int(input())
b=int(input())
try:
    ans=a/b
except:
    ans=a/2.0
print(ans)
```

```python
# try except finally block
try:
    a=int(input())
    b=int(input())
    c=a/b
    print("result:",c)
except ZeroDivisionError:
    print("you can not divide by zero")
except ValueError:
    print("please enter valid number")
finally:
    print("program executed with or without exception")
```

```python
#try except else block
try:
    a=int(input())
    b=int(input())
    c=a/b
    print("result:",c)
except ZeroDivisionError:
    print("you can not divide by zero")
except ValueError:
    print("please enter valid number")
else:
    print("program executed without exception")
```

```python
#normal if else program
age=int(input("please enter your age:"))
if age>=18:
    print("you are eligible to vote")
else:
    print("you are not eligible")
```
