```python
#abstract class-- a class whose object can not be created.
#it works as a blueprint for other class.
# it contains abstract method as well as normal methods and constructor
# it is used for partial abstraction
```

```python
from abc import ABC,abstractmethod
class Animal(ABC):
    @abstractmethod 
    def sound(self):
        pass
class Dog(Animal):
    def sound(self):
        print("Dog Barks")
class Cat(Animal):
    def sound(self):
        print("Cat meow")
d=Dog()
c=Cat()
d.sound()
c.sound()
```

```python
from abc import ABC,abstractmethod
class Bank(ABC):
    @abstractmethod
    def InterestRate(self):
        pass
class SBI(Bank):
    def InterestRate(self):
        print("in SBI,interest rate is 7%")
class HDFC(Bank):
    def InterestRate(self):
        print("in HDFC,interest rate is 6.5%")
s=SBI()
h=HDFC()
s.InterestRate()
h.InterestRate()
```

```python
#Interface - interface is not created using interface keyword like java,
#it is created using abstract method,it is a blueprint for the class
# it is used for fully abstraction
# it contains only abstract method
# it object can not be created
# it also allows to inherit multiple interface
# basically interface makes rules and classes follows rules
```

```python
from abc import ABC,abstractmethod
class Payment(ABC):
    @abstractmethod
    def pay(self):
        pass
class Stat(ABC):
    @abstractmethod
    def Status(self):
        pass
class PhonePay(Payment):
    def pay(self):
        print("payment made from PhonePay")
    def Status(self):
        print("Payment is successfull")
class PayTM(Payment):
    def pay(self):
        print("payment made from PayTM")
    def Status(self):
        print("Payment is unsuccessfull")
pp=PhonePay()
ptm=PayTM()
pp.pay()
pp.Status()
ptm.pay()

ptm.Status()
```
