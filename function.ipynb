```python
#arbitrary function
def EvenOdd(*numbers):
    EvenS=0
    OddS=0
    for i in numbers:
        if i%2==0:
            EvenS+=i
            print(i," is Even")
        else:
            OddS+=i
            print(i," is Odd")
    print("sum of all even:",EvenS)
    print("sum of all odd:",OddS)
```

```python
EvenOdd(1,2,3,4,5)
```

```python
def NumSum(n):
    if n==1:
        return 1
    else:
        return n+NumSum(n-1)
    
       
NumSum(1)
        
```

```python
def factorial(n):
    if n == 1:          # base case
        return 1
    else:
        return n * factorial(n-1)   # recursive call

#print(factorial(5))
factorial(5)
```
