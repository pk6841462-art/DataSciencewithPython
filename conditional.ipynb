```python
name=str(input("please, enter your name : "))
course=str(input("enter your course name : "))
maths=int(input("enter your marks in maths : "))
eng=int(input("enter your marks in english : "))
phy=int(input("enter your marks in physics : "))

total=(maths+eng+phy)
percent=(total/300)*100
print("total marks : ",total, "out of 300")
print("percentage : ",round(percent,2))

if (percent>=90):
    print("GRADE A")
    print("congratulations")
elif(percent>=70 and percent<90):
    print("GRADE B")
    print("excellent")
elif(percent>=35 and percent<70):
    print("GRADE C")
    print("very good")
else:
    print("FAIL\nplease work harder")
```
