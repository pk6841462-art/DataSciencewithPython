```python
class rectangle:
    def __init__(self,l,b):
        self.l=l
        self.b=b
    def area(self):
        print("area:",self.l*self.b)
    def peri(self):
        print("perimeter:",2*(self.l+self.b))
r1=rectangle(7,3)
r=rectangle(7,3)
del r
r.area()
r.peri()

r1.area()
r1.peri()
```

```python
class circle:
    def __init__(self,r):
        self.r=r
        self.pie=3.14
    def area(self):
        print("area:",self.pie*self.r*self.r)
    def perimeter(self):
        print("perimeter:",2*self.pie*self.r)
    def __del__(self):
        print(f"area:{(self.pie*self.r*self.r)} and perimeter:{(2*self.pie*self.r)}")
c=circle(1)
del c
c.area()
c.perimeter()
        
        
```
