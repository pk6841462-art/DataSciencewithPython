```python
class ParentClass:
    def __init__(self):
        print("this is parent class")
class ChildClass(ParentClass):
    def __init__(self):
        super().__init__()
        print("this is child class")
c=ChildClass()
```

```python
class animal:
    def __init__(self,name,color):
        self.name=name
        self.color=color
    def detail(self):
        print("name:",self.name)
        print("color:",self.color)
class dog(animal):
    def bark(self):
        print(f"{self.name} is barking")
d=dog("lebrador","black")
d.detail()
d.bark()

    
```

```python
class shape:
    def __init__(self,dim):
        self.dim=dim
    #setter method
    def setDim(self,dim):
        self.dim=dim
    #getter method
    def getDim(self):
        return self.dim
s=shape(2)
s.setDim(25)
s.getDim()
```
