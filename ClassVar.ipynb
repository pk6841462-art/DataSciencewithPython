```python
#definition of class variable
class student:
    school="arcade school"
    def __init__(self,name):
        self.name=name
        print(self.name)
        print(self.school)
obj=student("prashant")
```

```python
#how and where to declare class or static variable
#a. inside the class but outisde any method
class rectangle:
    color="blue"
    def __init__(self):
        self.l=1
        self.b=1
        print(self.l)
        print(self.b)
obj=rectangle()
print(rectangle.color)
```

```python
#b.inside class and inside constructor using class name
class rectangle:
    color="blue"
    def __init__(self):
        rectangle.age=23
        print(rectangle.age)
obj=rectangle()
print(obj.age)
print(rectangle.color)
```

```python
#c.inside class inside instance method using class name
class rectangle:
    def show(self):
        rectangle.name="prashant"
        print(rectangle.name)
        print(self.name)
obj=rectangle()
obj.show()
print(rectangle.name)
```

```python
#d.inside class,inside class method using cls or class name
class rec:
    @classmethod
    def method1(cls):
        cls.a=10
        rec.b=20
        print(rec.a)
        print(cls.b)
obj=rec()
obj.method1()
print(rec.a)
```

```python
#e.inside class inside static method using class name
class rec:
    @staticmethod
    def method1():
        rec.x=20
    @classmethod
    def method2(cls):
        print(rec.x)
obj=rec()
obj.method1()
obj.method2()
```

```python
#e.outside class using class name
class rec:
    pass
rec.school="arcade"
s1=rec()
s2=rec()
print(s1.school)
print(s2.school)
print(rec.school)
```

```python
college="amity"
class student:
    print("outside method but inside class:",college)
    def __init__(self):
        print(college)
    @classmethod
    def show(cls):
        cls.name="prashant"
        print("inside class and inside classmethod")
        print(cls.name)
        print(college)
    def dis():
        x=10
        print(x)
obj=student()
obj.show()
obj.dis()
        
```
