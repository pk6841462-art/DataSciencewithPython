```python
class student:
    def __init__(self,id=None,name=None):
        if id is None and name is None:
            self.id=0
            self.name="Default"
        else:
            self.id=id
            self.name=name
    def display(self):
        print(self.id,self.name)
```

```python
obj=student()
obj.display()
obj1=student(101,"prashant")
obj1.display()
obj2=student(102,'rajat')
obj2.display()
```

```python
class animal:
    def sound(self):
        print("animal makes sound")
class dog(animal):
    def sound(self):
        super().sound()
        print("dogs barks")
obj=dog()
```

```python
obj.sound()
```

```python
class wild:
    def __init__(self,name):
        self.name=name
class dog(wild):
    def __init__(self,name,breed):
        #self.name=name
        super().__init__(name)
        self.breed=breed
obj=dog("muffin","lebrador")
print(obj.name,obj.breed)
```

```python
class rectangle:
    def area(self,l,b):
        self.l=l
        self.b=b
        print("area:",l*b)
class square(rectangle):
    def area(self):
        super().area()
        print("area:",self.l*self.l)
obj=square()
obj.area(6,3)
```
