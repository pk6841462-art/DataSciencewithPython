```python
def outer_fun():
    #print("this is outer function")
    def inner_fun():
        print("this is inner join")
    return inner_fun()
```

```python
outer_fun()
```

```python
def calculate(*data):
    def add():
        total=0
        for i in data:
            total+=i
        print("sum:",total)
    def multiply():
        m=1
        for i in data:
            m*=i
        print("multiply:",m)
    add()
    multiply()
    
```

```python
calculate(1,2,3)
```

```python
#lambda function - function without its name
multiply=lambda a,b:a*b
```

```python
multiply(3,2)
```

```python
#pass by value
def increase(x):
    x=x+10
    print("inside method value of a is:",x)
a=5
increase(a)
print(a)
```

```python
#pass by reference
def modify(lst):
    lst.append(4)

numbers = [1,2,3]
print(numbers)
modify(numbers)
print(numbers)

```
