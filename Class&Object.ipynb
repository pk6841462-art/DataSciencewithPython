```python
class rectangle:
    def __init__(self):
        self.l=2
        self.b=3
    def dis(self):
        print("length:",self.l)
        print("breadth:",self.b)
    def col(self):
        self.color="red"
        print(self.color)
obj=rectangle()
print(obj.__dict__)
obj.dis()
obj.col()
```

```python
class cuboid:
    def __init__(self,l,b,h):
        self.l=l
        self.b=b
        self.h=h
    def surfaceArea(self):
        return (self.l*self.b*self.h)
    def volume(self):
        return 2*(self.l*self.b+self.b*self.h+self.h*self.l)
obj=cuboid(2,3,4)
print("area:",obj.surfaceArea())
print("volume:",obj.volume())
```
